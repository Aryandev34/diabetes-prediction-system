# Diabetes Prediction System
Binary classification with **SVM** — Pandas, NumPy, Scikit-Learn.

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
```

```python
dataset = pd.read_csv('../data/diabetes.csv')
dataset.head()
```

```python
print(dataset.shape)
print(dataset['Outcome'].value_counts())
dataset.groupby('Outcome').mean(numeric_only=True)
```

## Features, standardization & stratified split

```python
X = dataset.drop(columns='Outcome')
y = dataset['Outcome']
scaler = StandardScaler()
X = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=2)
print(X.shape, X_train.shape, X_test.shape)
```

## Train SVM & evaluate accuracy

```python
classifier = SVC(kernel='linear')
classifier.fit(X_train, y_train)
print('Training accuracy:', accuracy_score(y_train, classifier.predict(X_train)))
print('Testing accuracy:', accuracy_score(y_test, classifier.predict(X_test)))
```

## Automated prediction for a new patient

```python
input_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)
std = scaler.transform(np.asarray(input_data).reshape(1, -1))
pred = classifier.predict(std)
print('The person is Diabetic' if pred[0] == 1 else 'The person is not Diabetic')
```
